# Data Cleaning


```python
df_clean = df.drop_duplicates().copy()
```

### filling missing values


```python
df_clean["Price"] = df_clean["Price"].fillna(df_clean["Price"].median())
df_clean["Quantity"] = df_clean["Quantity"].fillna(df_clean["Quantity"].median())
```

```python
df_clean["Revenue"] = df_clean["Price"] * df_clean["Quantity"]
```

```python
df_clean["Total_Cost"] = (
    df_clean["Cost_Per_Unit"] * df_clean["Quantity"]
)
```

```python
df_clean["Profit"] = (
    df_clean["Revenue"] - df_clean["Discount"] - df_clean["Total_Cost"]
)
```

```python
df_clean["Profit_Margin_%"] = np.where(
    df_clean["Revenue"] != 0,
    df_clean["Profit"] / df_clean["Revenue"] * 100,
    np.nan
)
```

```python
display(df_clean)
```

```python
print(df_clean.isnull().sum())
```

```python
print("Duplicate rows:", df_clean.duplicated().sum())
```

```python
print("Total revenue:", df_clean["Revenue"].sum())
print("Total profit:", df_clean["Profit"].sum())
print("Average profit margin:", df_clean["Profit_Margin_%"].mean())
```

# Applying different filters

### 1. Filter by category

```python
print("Electronics transactions:")
display(df_clean[df_clean["Category"] == "Electronics"])
```

### 2. Quantity greater than 5

```python
print("Transactions with quantity above 5:")
display(df_clean[df_clean["Quantity"] > 5])
```

### 3. AND condition: Chennai and revenue above 5000

```python
print("Chennai transactions with revenue above 5000:")
display(df_clean[
    (df_clean["Region"] == "Chennai") &
    (df_clean["Revenue"] > 5000)
])
```

### 4. OR condition: Chennai or Bengaluru

```python
print("Transactions from Chennai or Bengaluru:")
display(df_clean[
    (df_clean["Region"] == "Chennai") |
    (df_clean["Region"] == "Bengaluru")
])
```

### 5. Date range: October 2 to October 4

```python
print("Transactions from October 2 to October 4:")
display(df_clean[
    (df_clean["Date"] >= "2026-10-02") &
    (df_clean["Date"] <= "2026-10-04")
])
```

### 6. Filter by derived column: profit above 5000

```python
print("Transactions with profit above 5000:")
display(df_clean[df_clean["Profit"] > 5000])
```

### 7. NOT condition: exclude Electronics

```python
print("Transactions excluding Electronics:")
display(df_clean[~(df_clean["Category"] == "Electronics")])
```
