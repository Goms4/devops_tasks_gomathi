### converting columns to appropriate data types

```python
print("INFO BEFORE CONVERSION")
df.info()

missing_before = df.isnull().sum()
```

```python
df["UnitPrice"] = (
    df["UnitPrice"]
    .astype("string")
    .str.replace("$", "", regex=False)
    .str.strip()
)
```

```python
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["UnitPrice"] = pd.to_numeric(df["UnitPrice"], errors="coerce")
```

```python
df["OrderDate"] = pd.to_datetime(df["OrderDate"], errors="coerce")
```

Standardize inconsistent strings

```python
string_columns = [
    "CustomerName", "City", "Product", "PaymentMethod"
]

for col in string_columns:
    df[col] = df[col].astype("string").str.strip().str.title()

print("String standardization completed.")
display(df.head(10))
```

```python
print("Rows containing missing values:")
display(df[df.isnull().any(axis=1)])
```

```python
df = df.dropna(subset=["OrderDate"]).copy()
```

```python
for col in ["CustomerName", "City", "Product", "PaymentMethod"]:
    df[col] = df[col].fillna("Unknown")
```

```python
for col in ["Quantity", "UnitPrice"]:
    df[col] = df[col].fillna(df[col].median())
```

```python
df["Quantity"] = df["Quantity"].astype("int64")
```

```python
print(df.isnull().sum())
```

```python
df.info()
```

```python
Q1 = df["UnitPrice"].quantile(0.25)
Q3 = df["UnitPrice"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["UnitPrice"] < lower_bound) |
    (df["UnitPrice"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

print("\nPotential price outliers:")
display(outliers[["OrderID", "Product", "UnitPrice"]])
```

visualize the outliers

```python
df[["UnitPrice"]].plot(
    kind="box",
    figsize=(7, 5),
    title="Box Plot of E-commerce Unit Prices"
)

plt.ylabel("Unit Price")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

handling outliers

```python
df["UnitPrice_Original"] = df["UnitPrice"]

df["UnitPrice"] = df["UnitPrice"].clip(
    lower=lower_bound,
    upper=upper_bound
)

print("Outlier handling completed.")
```

creating derived features

```python
df["TotalAmount"] = df["Quantity"] * df["UnitPrice"]
```

# Feature engineering using apply()

```python
def categorize_order(amount):
    if amount >= 50000:
        return "High Value"
    elif amount >= 10000:
        return "Medium Value"
    else:
        return "Low Value"

df["OrderCategory"] = df["TotalAmount"].apply(categorize_order)

display(df[[
    "OrderID", "Product", "Quantity",
    "UnitPrice", "TotalAmount", "OrderCategory"
]].head(10))
```
